```python
import pandas as pd
```

```python
meta_score = pd.read_excel("C:/Users/Emma/Desktop/metaClean43Brightspace.xlsx")
```

```python
meta_score.head()
```

```python
#splits the string values into lists separated by comma
meta_score["cast"] = meta_score["cast"].str.split(",")
meta_score["awards"] = meta_score["awards"].str.split(",")
```

```python
meta_score[["cast", "awards"]]
```

```python
#creates a new data frame for actors overview
actors = pd.DataFrame(columns = ["actor_id", "actor_name"])
```

```python
#imports data from meta score df to actors df
actors["actor_name"] = meta_score["cast"]
actors = actors.explode("actor_name")
#normalizes the data
actors["actor_name"] = actors["actor_name"].astype(str)
actors["actor_name"] = actors["actor_name"].str.lower()
```

```python
actors.head(20)
```

```python
#creates actor id based on index to fill in NA cells to later 
actors["actor_id"] = actors.index + 1
```

```python
#removes duplicates
actors.reset_index(drop=True, inplace=True)
actors["actor_name"] = actors["actor_name"].drop_duplicates()
actors.head()
```

```python
#removes NA values
actors.reset_index(drop=True, inplace=True)
actors = actors.dropna()
```

```python
actors.head(20)
```

```python
#check for duplicates
print(actors["actor_name"].duplicated().sum())
```

```python
#alphabatically sorts the names
actors = actors.sort_values(by = "actor_name")
```

```python
actors
```

```python
#resets the index after sorting
actors.reset_index(drop=True, inplace=True)
actors["actor_id"] = actors.index + 1
```

```python
actors.head()
```

```python
actors = actors.drop(index=0)
```
