```python
import pandas as pd
from sqlalchemy import create_engine, text, inspect
```

```python
db_url = "postgresql+psycopg://postgres:1234@localhost:5432/movies_db"
engine = create_engine(db_url)
```

```python
def get_highest_movie_revenue(engine):
    query = """select m.movie_id,
        m.movie_name,
        s.production_budget,
        s.total_revenue,
        s.total_revenue - s.production_budget as profit
    from movie_table m
    left join sales_table s
    on m.movie_id = s.movie_id
    where s is not null
    order by s.total_revenue desc"""
    return pd.read_sql(query, engine)
```

```python
get_highest_movie_revenue(engine)
```
