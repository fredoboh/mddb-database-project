```python
import pandas as pd
```

```python
meta_score = pd.read_excel("C:/Users/Emma/Desktop/metaClean43Brightspace.xlsx")
```

```python
#splits the string values into lists separated by comma
meta_score["cast"] = meta_score["cast"].str.split(",")
meta_score["awards"] = meta_score["awards"].str.split(",")
```

```python
#creates a new data frame for directors overview
directors = pd.DataFrame(columns = ["director_id", "director_name"])
```

```python
#imports data from meta score df to directors df
directors["director_name"] = meta_score["director"]
#normalises the data entries
directors["director_name"] = meta_score["director"].astype(str)
directors["director_name"] = meta_score["director"].str.lower()
```

```python
#creates director id based on index to fill in NA cells to later 
directors["director_id"] = directors.index + 1
```

```python
#removes duplicates from the dataframe
directors.reset_index(drop=True, inplace=True)
directors["director_name"] = directors["director_name"].drop_duplicates()
```

```python
#removes empty values left after the removal of the duplicates
directors.reset_index(drop=True, inplace=True)
directors = directors.dropna()
```

```python
#duplicate check
print(directors.duplicated().sum())
```

```python
#alphabatically sorts the names
directors = directors.sort_values(by = "director_name")
```

```python
#resets the index after sorting to create an id values
directors.reset_index(drop=True, inplace=True)
directors["director_id"] = directors.index + 1
```
