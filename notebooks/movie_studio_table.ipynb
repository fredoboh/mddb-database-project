```python
import pandas as pd
```

```python
meta_score = pd.read_excel("C:/Users/Emma/Desktop/metaClean43Brightspace.xlsx")
movie_table = pd.read_csv("C:/Users/Emma/Downloads/movies_table.csv")
studio_table = pd.read_csv("C:/Users/Emma/Downloads/studios.csv")
```

```python
#splits the string values into lists separated by comma
meta_score["cast"] = meta_score["cast"].str.split(",")
meta_score["awards"] = meta_score["awards"].str.split(",")
```

```python
#creates new dataframe for the relationship table
movie_studio = pd.DataFrame(columns = ["url", "studio_name"])
```

```python
movie_studio["url"] = meta_score["url"]
movie_studio["studio_name"] = meta_score["studio"]
```

```python
#normalises the data entries
movie_studio["studio_name"] = movie_studio["studio_name"].astype(str)
movie_studio["studio_name"] = movie_studio["studio_name"].str.lower()
```

```python
movie_studio.head()
```

```python
#merges the combined table movie_director with director table based on the director name
movie_studio = movie_studio.merge(studio_table, on="studio_name")
```

```python
movie_studio = movie_studio.drop("studio_name", axis=1)
```

```python
movie_studio.head()
```

```python
movie_studio = movie_studio.merge(movie_table, on="url")
```

```python
movie_studio.head()
```

```python
#removes the unnecessary columns
movie_studio = movie_studio.drop(columns=['url', 'movie_name', 'RelDate', 'runtime', 'rating','summary', 'userscore', 'metascore'])
```

```python
#sorts the movie_director table asc based on movie id
movie_studio = movie_studio.sort_values(by = "movie_id")
```

```python
movie_studio
```

```python
#resets the index and creates a new id for movie-director combination
movie_studio.reset_index(drop=True, inplace=True)
movie_studio["movie_studio_id"] = movie_studio.index + 1
```
