```python
#Creating a function to clean the movie names and remove duplicates
def remove_duplicates(movie_names):
    """converts the movie names to lowercase, removes any text in parentheses or brackets, 
    removes punctuation, and sorts the words in the movie name. 
    It then checks if the sorted movie name is already in a set of seen movie names. If it is not, 
    it adds the sorted movie name to the set and appends the original cleaned movie name to a list of unique movie names. 
    Finally, it returns the list of unique movie names in"""
    seen = {}
    unique_movie_names = []
    stop_words = {"the", "a", "an", "of", "and", "in", "to", "for"}
    for index, movie_name in movie_names.items():
        movie_name = str(movie_name).lower()
        cleaned = re.sub(r"\s*[\(\[].*[a-zA-Z].*[\)\]]", "", movie_name)
        cleaned = re.sub(r"[^\w\s]", "", cleaned)
        sort_movie_name = tuple(sorted([word for word in cleaned.split() if word not in stop_words]))
        if sort_movie_name not in seen:
            seen[sort_movie_name] = len(unique_movie_names)
            unique_movie_names.append((index, cleaned))
        else:
            position = seen[sort_movie_name]
            unique_movie_names[position] = (index, cleaned)
    return unique_movie_names
```

```python
def remove_duplicates(movie_names):
    seen = {}
    unique_movie_names = []
    stop_words = {"the", "a", "an", "of", "and", "in", "to", "for"}

    for index, movie_name in movie_names.items():
        movie_name = str(movie_name).lower()
        cleaned = re.sub(r'\s*[\(\[].*[a-zA-Z].*[\)\]]', '', movie_name)
        cleaned = re.sub(r'[^\w\s]', '', cleaned)

        sort_movie_name = tuple(
            sorted(word for word in cleaned.split() if word not in stop_words)
        )

        if sort_movie_name not in seen:
            seen[sort_movie_name] = len(unique_movie_names)
            unique_movie_names.append((index, cleaned))
        else:
            # Replace the first occurrence with the second
            position = seen[sort_movie_name]
            unique_movie_names[position] = (index, cleaned)
        

    return unique_movie_names
```

```python
#Cleaning the movie names and removing duplicates
cleaned_movie_names = remove_duplicates(urls_df["movie_name"])
cleaned_movie_names
```

```python
#Assigning the indexes and titles to urls_df
indexes, titles = zip(*cleaned_movie_names)
urls_df = urls_df.loc[list(indexes)].copy()
urls_df["movie_name"] = titles
display(urls_df.shape)
display(urls_df.head())
```

```python
#Creating a function to convert date to datetime
def parse_date(date):
    """Converts messy dates to datetime and turns pure strings or non date vales to NaT"""
    try:
        return parser.parse(str(date))
    except (parser.ParserError, TypeError, ValueError):
        return pd.NaT
```

```python
#Converting the "RelDate" column to datetime format, coercing errors to NaT for invalid dates
movies["RelDate"] = movies["RelDate"].apply(parse_date)
```

```python
#Assigning a unique movie_id to each movie based on the index of the dataframe and creating my movies table with the required columns
movies["movie_id"] = movies.index + 1
movies_table = movies[["movie_id","movie_name", "RelDate", "runtime", "rating", "summary", "userscore", "metascore", "url"]].copy()
```

```python
#Extracting the required columns
user_review = df3[["url", "idvscore", "reviewer", "dateP", "Rev", "thumbsUp", "thumbsTot", "Tone"]].copy()
```

```python
user_review["title"] = user_review["url"].str.split("/", expand=True).iloc[:,-1]
```

```python
stop_words = {"the", "a", "an", "of", "and", "in", "to", "for"}
def normalized_movie_key(name, stopwords):
    name = name.lower()
    name = re.sub(r'\s*[\(\[].*[a-zA-Z].*[\)\]]', "", name)
    name = name.replace("+", "")
    name = re.sub(r"[^\w\s]", " ", name)
    words = sorted([word for word in name.split() if word not in stop_words])
    return " ".join(words)
```

```python
user_review["movie_key"] = user_review["title"].apply(lambda x: normalized_movie_key(x, stopwords=stop_words))
movies_table["movie_key"] = movies_table["movie_name"].apply(lambda x: normalized_movie_key(x, stopwords=stop_words))
```

```python
user_review_m = user_review.merge(movies_table[["movie_key", "movie_id"]], on= "movie_key", how= "left")
```

```python
user_review_m.info()
```

```python
user_review_m = user_review_m.drop(columns= ["movie_key", "title"]).rename(columns= {"idvscore" : "indv_score",
                                                                                    "dateP" : "date_posted",
                                                                                    "Rev" : "review",
                                                                                    "thumbsUp" : "likes",
                                                                                    "thumbsTot" : "dislikes",
                                                                                    "Tone" : "tone"})
```

```python
user_review_m["reviewer"] = user_review_m["reviewer"].str.replace("'", "").str.lower()
user_review_m["date_posted"] = user_review_m["date_posted"].str.replace("'", "").str.strip()
user_review_m["review"] = user_review_m["review"].str.replace("'", "").str.lower()
```

```python
user_review_m["date_posted"] = user_review_m["date_posted"].apply(parse_date)
user_review_m["likes"] = pd.to_numeric(user_review_m["likes"], errors= "coerce").astype("Int64")
user_review_m["dislikes"] = pd.to_numeric(user_review_m["dislikes"], errors = "coerce").astype("Int64")
user_review_m["indv_score"] = pd.to_numeric(user_review_m["indv_score"], errors="coerce")
```
