```python
import pandas as pd
```

```python
meta_score = pd.read_excel("C:/Users/Emma/Desktop/metaClean43Brightspace.xlsx")
```

```python
#splits the string values into lists separated by comma
meta_score["cast"] = meta_score["cast"].str.split(",")
meta_score["awards"] = meta_score["awards"].str.split(",")
```

```python
#creates a new data frame for studio overview
studios = pd.DataFrame(columns = ["studio_id", "studio_name"]) 
```

```python
#imports data from meta score df to directors df
studios["studio_name"] = meta_score["studio"]
#normalises the data entries
studios["studio_name"] = studios["studio_name"].astype(str)
studios["studio_name"] = studios["studio_name"].str.lower()
```

```python
#creates actor id based on index to fill in NA cells to later 
studios["studio_id"] = studios.index +1
```

```python
#removes duplicates from the dataframe
studios.reset_index(drop=True, inplace=True)
studios["studio_name"] = studios["studio_name"].drop_duplicates()
```

```python
#removes NA values from the dataframe
studios.reset_index(drop=True, inplace=True)
studios = studios.dropna()
```

```python
#duplicate check
print(studios.duplicated().sum())
```

```python
#alphabatically sorts the names
studios = studios.sort_values(by = "studio_name")
```

```python
#resets the index after sorting to create an id values
studios.reset_index(drop=True, inplace=True)
studios["studio_id"] = studios.index + 1
```

```python
studios["studio_names"] = studios["studio_name"].replace('"', regex=False)
```
