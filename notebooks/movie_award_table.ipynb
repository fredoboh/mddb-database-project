```python
import pandas as pd
```

```python
meta_score = pd.read_excel("C:/Users/Emma/Desktop/metaClean43Brightspace.xlsx")
movie_table = pd.read_csv("C:/Users/Emma/Downloads/movies_table.csv")
award_table = pd.read_csv("C:/Users/Emma/Downloads/awards.csv")
```

```python
#splits the string values into lists separated by comma
meta_score["cast"] = meta_score["cast"].str.split(",")
meta_score["awards"] = meta_score["awards"].str.split(",")
```

```python
#creates new dataframe for the relationship table
movie_award = pd.DataFrame(columns = ["url", "award_name"])
```

```python
movie_award["url"] = meta_score["url"]
movie_award["award_name"] = meta_score["awards"]
```

```python
#expands the created lists into rows
movie_award = movie_award.explode("award_name")
#normalises the data entries
movie_award["award_name"] = movie_award["award_name"].astype(str)
movie_award["award_name"] = movie_award["award_name"].str.lower()
```

```python
movie_award.head()
```

```python
#merges the combined table movie_award with actor table based on the award name
movie_award = movie_award.merge(award_table, on="award_name")
```

```python
movie_award = movie_award.drop("award_name", axis=1)
```

```python
movie_award.head()
```

```python
#merges the combined table movie_award with movie table based on the url
movie_award = movie_award.merge(movie_table, on="url")
```

```python
#removes the unnecessary columns
movie_award = movie_award.drop(columns=['url','movie_name', 'RelDate', 'runtime', 'rating','summary', 'userscore', 'metascore'])
```

```python
#sorts the movie_actor table asc based on movie id
movie_award = movie_award.sort_values(by = "movie_id")
```

```python
movie_award
```

```python
#removes unnecessary columns
movie_award = movie_award.drop(columns=['award_body','category'])
```

```python
#resets the index and creates a new id for movie-actor combination
movie_award.reset_index(drop=True, inplace=True)
movie_award["movie_award_id"] = movie_award.index + 1
```
