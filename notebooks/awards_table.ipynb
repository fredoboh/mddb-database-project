```python
import pandas as pd
```

```python
meta_score = pd.read_excel("C:/Users/Emma/Desktop/metaClean43Brightspace.xlsx")
```

```python
#splits the string values into lists separated by comma
meta_score["cast"] = meta_score["cast"].str.split(",")
meta_score["awards"] = meta_score["awards"].str.split(",")
```

```python
meta_score.head()
```

```python
#creates a new data frame for awards overview
awards = pd.DataFrame(columns = ["award_id", "award_name","award_body","category"])
```

```python
##imports data from meta score df to actors df
awards["award_name"] = meta_score["awards"]
awards = awards.explode("award_name")
#normalises the data entries
awards["award_name"] = awards["award_name"].astype(str)
awards["award_name"] = awards["award_name"].str.lower()
```

```python
awards["award_body"] = awards["award_body"].fillna("-")
awards["category"] = awards["category"].fillna("-")
awards.head()
```

```python
#creates award id based on index to fill in NA cells to later 
awards["award_id"] = awards.index + 1
awards.head()
```

```python
#removes duplicates from the dataframe
awards.reset_index(drop=True, inplace=True)
awards["award_name"] = awards["award_name"].drop_duplicates()
```

```python
#removes NA values from the dataframe
awards.reset_index(drop=True, inplace=True)
awards = awards.dropna()
```

```python
#alphabatically sorts the names
awards = awards.sort_values(by = "award_name")
```

```python
#resets the index after sorting to create an id values
awards.reset_index(drop=True, inplace=True)
awards["award_id"] = awards.index + 1
```
