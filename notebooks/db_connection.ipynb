```python
from dateutil import parser
def parse_date(date):
    try:
        return parser.parse(str(date))
    except (parser.ParserError, TypeError, ValueError):
        return pd.NaT
```

```python
movies["release_date"] = movies["release_date"].apply(parse_date)
```
