```python
import pandas as pd
```

```python
meta_score = pd.read_excel("C:/Users/Emma/Desktop/metaClean43Brightspace.xlsx")
movie_table = pd.read_csv("C:/Users/Emma/Downloads/movies_table.csv")
director_table = pd.read_csv("C:/Users/Emma/Downloads/directors.csv")
```

```python
#creates new dataframe for the relationship table
movie_director = pd.DataFrame(columns = ["url", "director_name"])
```

```python
movie_director["url"] = meta_score["url"]
movie_director["director_name"] = meta_score["director"]
```

```python
#normalises the data enrty
movie_director["director_name"] = movie_director["director_name"].astype(str)
movie_director["director_name"] = movie_director["director_name"].str.lower()
```

```python
director_table
```

```python
#merges the combined table movie_director with director table based on the director name
movie_director = movie_director.merge(director_table, on="director_name")
```

```python
movie_director = movie_director.drop("director_name", axis=1)
```

```python
movie_director.head()
```

```python
#merges the combined table movie_director with movie  table based on the url
movie_director = movie_director.merge(movie_table, on="url")
```

```python
#removes the unnecessary columns
movie_director = movie_director.drop(columns=['url', 'movie_name', 'RelDate', 'runtime', 'rating','summary', 'userscore', 'metascore'])
```

```python
movie_director
```

```python
#sorts the movie_director table asc based on movie id
movie_director = movie_director.sort_values(by = "movie_id")
```

```python
movie_director
```

```python
#resets the index and creates a new id for movie-director combination
movie_director.reset_index(drop=True, inplace=True)
movie_director["movie_director_id"] = movie_director.index + 1
```
