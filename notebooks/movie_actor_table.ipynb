```python
import pandas as pd
```

```python
meta_score = pd.read_excel("C:/Users/Emma/Desktop/metaClean43Brightspace.xlsx")
movie_table = pd.read_csv("C:/Users/Emma/Downloads/movies_table.csv")
actor_table = pd.read_csv("C:/Users/Emma/Downloads/actors.csv")
```

```python
meta_score.head()
```

```python
#splits the string values into lists separated by comma
meta_score["cast"] = meta_score["cast"].str.split(",")
meta_score["awards"] = meta_score["awards"].str.split(",")
```

```python
#creates new dataframe for the relationship table
movie_actor = pd.DataFrame(columns = ["url", "actor_name"])
```

```python
movie_actor["url"] = meta_score["url"]
movie_actor["actor_name"] = meta_score["cast"]
```

```python
movie_actor.head()
```

```python
#expands the created lists into rows
movie_actor = movie_actor.explode("actor_name")
#normalises the data entires
movie_actor["actor_name"] = movie_actor["actor_name"].astype(str)
movie_actor["actor_name"] = movie_actor["actor_name"].str.lower()
```

```python
movie_actor.head()
```

```python
#merges the combined table movie_actor with actor table based on the actor name
movie_actor = movie_actor.merge(actor_table, on="actor_name")
```

```python
movie_actor.head()
```

```python
movie_actor = movie_actor.drop("actor_name", axis=1)
```

```python
#merges the combined table movie_actor with movie  table based on the url
movie_actor = movie_actor.merge(movie_table, on="url")
```

```python
#removes the unnecessary columns
movie_actor = movie_actor.drop(columns=['url', 'movie_name', 'RelDate', 'runtime', 'rating','summary', 'userscore', 'metascore'])
```

```python
movie_actor
```

```python
#sorts the movie_actor table asc based on movie id
movie_actor= movie_actor.sort_values(by = "movie_id")
```

```python
movie_actor.head()
```

```python
#resets the index and creates a new id for movie-actor combination
movie_actor.reset_index(drop=True, inplace=True)
movie_actor["movie_actor_id"] = movie_actor.index + 1
```
